```python
import pandas as pd

books = pd.read_csv(f"archive/books.csv")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ax = plt.axes()
sns.heatmap(books.isna().transpose(), cbar=False, ax=ax)

plt.xlabel("Columns")
plt.ylabel("Missing Values")
```

```python
import numpy as np

books["missing_description"] = np.where(books["description"].isna(), 1, 0)
books["age_of_book"] = 2024 - books["published_year"]
```

```python
columns_of_interest = [
    "num_pages",
    "missing_description",
    "age_of_book",
    "average_rating",
]
correlational_matrix = books[columns_of_interest].corr(method="spearman")

sns.set_theme(style="white")
plt.figure(figsize=(8, 6))

heatmap = sns.heatmap(
    correlational_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    square=True,
    linewidths=0.5,
    cbar_kws={"label": "Spearman correlation"},
)

heatmap.set_title("Spearman Correlation Heatmap", fontsize=16)
plt.show()
```

```python
books_missing = books[
    ~(books["description"].isna())
    & ~(books["num_pages"].isna())
    & ~(books["average_rating"].isna())
    & ~(books["published_year"].isna())
]
```

```python
books_missing["categories"].value_counts().reset_index().sort_values(
    "count", ascending=False
)
```

```python
books_missing["words_in_description"] = (
    books_missing["description"].str.split().str.len()
)
```

```python
books_missing
```

```python
books_missing.loc[books_missing["words_in_description"].between(1, 4), "description"]
```

```python
book_missing_25_words = books_missing[books_missing["words_in_description"] >= 25]
```

```python
book_missing_25_words
```

```python
book_missing_25_words["title_and_subtitle"] = (
    book_missing_25_words["title"]
    .str.cat(book_missing_25_words["subtitle"], sep=": ")
    .fillna(book_missing_25_words["title"])
)
```

```python
book_missing_25_words["tagged_description"] = (
    book_missing_25_words["isbn13"]
    .astype(str)
    .str.cat(book_missing_25_words["description"].astype(str), sep=": ")
)
```

```python
(
    book_missing_25_words.drop(
        ["subtitle", "missing_description", "age_of_book", "words_in_description"],
        axis=1,
    ).to_csv("archive/books_cleaned.csv", index=False)
)
```
